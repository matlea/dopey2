## Fermi map

### Load data

```python
data = dopey.load("../data_for_dev/fermi_map.xy")
```

### Interactive viewer

```python
dopey.plot(data)
```

### Cut data

```python
cutXY = dopey.subArray(data, axis = 2, a1 = 7.52, a2 = 7.54)
cutXY = dopey.compact(cutXY, axis = 2)

cutYE = dopey.subArray(data, axis = 0, a1 = -0.5, a2 = 0.5)
cutYE = dopey.compact(cutYE, axis = 0)

cutXE = dopey.subArray(data, axis = 1, a1 = -0.5, a2 = 0.5)
cutXE = dopey.compact(cutXE, axis = 1)

fig, ax = plt.subplots(ncols = 3, figsize = (9,3))
ax[0] = dopey.plot(cutXY, ax = ax[0], transpose = True)
ax[1] = dopey.plot(cutYE, ax = ax[1], transpose = True)
ax[2] = dopey.plot(cutXE, ax = ax[2], transpose = True)
fig.tight_layout()
```

```python
E1, E2 = 7.53, 7.55
curve = dopey.subArray(cutYE, axis = 1, a1 = E1, a2 = E2)
curve = dopey.compact(curve, axis = 1)

fig, ax = plt.subplots(ncols = 2, figsize = (6,3))
ax[0] = dopey.plot(cutYE, ax = ax[0], transpose = True)
for E in [E1, E2]: ax[0].axhline(y = E, linewidth = 0.6, color = "k")
ax[1] = dopey.plot(curve, ax = ax[1])
fig.tight_layout()
```
